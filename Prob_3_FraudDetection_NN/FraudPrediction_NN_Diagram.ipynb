```python
#Import the required Libraries to draw the NN
import networkx as nx
import matplotlib.pyplot as plt
```

```python
#In order to draw the NN, need to define the NN with the parameters
#Input Layer
input_features = ["TransactionAmount","TransactionTime","MerchantCategory",
"CustomerAge","AccountBalance","NumberOfTransactionsToday"]

#Hidden Layer as we have only 6 input features for this model diagram am using only two Hidden Layer
hidden_layer1 = [f"HN{i}" for i in range(1,9)]
hidden_layer2 = [f"HL{i}" for i in range(1,7)]
#Output Layer 
output_layer = ["Fraud-Prediction"]
```

```python
#Create graph using Networkx library

#First create empyt Directed graph object
G = nx.DiGraph()

#Add nodes for (Input, Hidden, Output)
G.add_nodes_from(input_features)
G.add_nodes_from(hidden_layer1)
G.add_nodes_from(hidden_layer2)
G.add_nodes_from(output_layer)

#Add edge(-> from Input Layer to the Hidden Layer)
for i in input_features:
    for h in hidden_layer1:
        G.add_edge(i, h)
#Add edge(-> from Hidden Layer1 to the Hidden Layer2)
for i in hidden_layer1:
    for h in hidden_layer2:
        G.add_edge(i, h)

#Add edge(-> from Hidden layer2 to Output layer)
for i in hidden_layer2:
    G.add_edge(i, output_layer[0])

```

```python
#Now position the nodes
pos = {}
#Position for Input Layer
for i,feature in enumerate(input_features):
    pos[feature] = (0, i)
#Position for Hidden Layer1
for i, neuron in enumerate(hidden_layer1):
    pos[neuron] = (1.5, i * 0.7)
#Position for Hidden Layer2
for i, neuron in enumerate(hidden_layer2):
    pos[neuron] = (3, i * 0.7)
#Position for the Output Layer
pos[output_layer[0]] = (4,2)
```

```python
#Now draw the graph
plt.figure(figsize=(14,8))
nx.draw(G, pos, with_labels= True,node_size=1500, node_color="lightgreen",
        arrowsize=15, font_size=12)
plt.title("Neural Network Achitecture for Fraud Detection")
plt.show()
```
